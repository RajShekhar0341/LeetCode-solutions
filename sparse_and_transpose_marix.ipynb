```python
# This code was originally in C++ and has been converted to Python.

m = int(input("Enter m:"))
n = int(input("Enter n:"))

arr = []
for r in range(m):
    row = []
    for c in range(n):
        user_input = int(input(f"Enter input for element ({r},{c}): "))
        row.append(user_input)
    arr.append(row)

print("\nYour matrix:")
for r in range(m):
    for c in range(n):
        print(arr[r][c], end=" ")
    print()
```

```python
#include <iostream>
#include <vector>

int main() {
    int m, n;

    std::cout << "Enter m: ";
    std::cin >> m;
    std::cout << "Enter n: ";
    std::cin >> n;

    std::vector<std::vector<int>> arr(m, std::vector<int>(n));

    for (int r = 0; r < m; ++r) {
        for (int c = 0; c < n; ++c) {
            std::cout << "Enter input for element (" << r << "," << c << "): ";
            std::cin >> arr[r][c];
        }
    }

    std::cout << "\nYour matrix:\n";
    for (int r = 0; r < m; ++r) {
        for (int c = 0; c < n; ++c) {
            std::cout << arr[r][c] << " ";
        }
        std::cout << std::endl;
    }

    return 0;
}
```
